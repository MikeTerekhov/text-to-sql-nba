# Using sqlite3, run basic query on nba database

```python
import sqlite3 as sql

connection = sql.connect('./nba-data/nba.sqlite')
cursor = connection.cursor()

cursor.execute("select * from player limit 5")
rows = cursor.fetchall()
for row in rows:
    print(row)

print('\n\n')
cursor.execute("select * from player where first_name = 'LeBron'")
rows = cursor.fetchall()
for row in rows:
    print(row)
```
